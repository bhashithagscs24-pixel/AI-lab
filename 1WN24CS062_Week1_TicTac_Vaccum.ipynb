```python
#week1
# tic-tac-toe

import random

# Initialize the board
board = [" " for _ in range(10)]

# Path cost
path_cost = 0


def print_board(brd):
    print(brd[1] + "|" + brd[2] + "|" + brd[3])
    print("-+-+-")
    print(brd[4] + "|" + brd[5] + "|" + brd[6])
    print("-+-+-")
    print(brd[7] + "|" + brd[8] + "|" + brd[9])


def is_winner(brd, letter):
    winning_combinations = [
        [1, 2, 3], [4, 5, 6], [7, 8, 9],
        [1, 4, 7], [2, 5, 8], [3, 6, 9],
        [1, 5, 9], [3, 5, 7]
    ]

    return any(
        all(brd[i] == letter for i in combo)
        for combo in winning_combinations
    )


def is_board_full(brd):
    return brd.count(" ") == 1


def player_move():
    global path_cost

    run = True

    while run:
        move = input("Select a position for 'O' (1-9): ")

        try:
            move = int(move)

            if 1 <= move <= 9:

                if board[move] == " ":
                    board[move] = "O"

                    # Path cost for player's move
                    path_cost += 1

                    print("Player move cost = 1")
                    print("Current path cost =", path_cost)

                    run = False

                else:
                    print("This space is already occupied!")

            else:
                print("Please type a number between 1 and 9.")

        except ValueError:
            print("Please type a valid number.")


def computer_move():
    global path_cost

    # Find all empty spaces
    empty_indexes = [
        x for x, letter in enumerate(board)
        if letter == " " and x != 0
    ]

    move = 0

    # Check if computer can win
    # or needs to block player
    for let in ["X", "O"]:

        for i in empty_indexes:

            copy_board = list(board)
            copy_board[i] = let

            if is_winner(copy_board, let):
                move = i

                assign_move(move)

                # Path cost for computer move
                path_cost += 1

                print("Computer move cost = 1")
                print("Current path cost =", path_cost)

                return

    # Otherwise pick a random valid move
    if empty_indexes:

        move = random.choice(empty_indexes)

        assign_move(move)

        # Path cost for computer move
        path_cost += 1

        print("Computer move cost = 1")
        print("Current path cost =", path_cost)


def assign_move(move):
    board[move] = "X"


def play_game():
    global path_cost

    print("Welcome to Tic Tac Toe!")

    # Initial board
    board[1] = "O"
    board[3] = "X"
    board[4] = "X"
    board[7] = "X"
    board[8] = "O"
    board[9] = "O"

    # Initial moves already present
    path_cost = 6

    print_board(board)

    print("\nInitial Path Cost =", path_cost)

    while not is_board_full(board):

        # Computer's turn
        if not is_winner(board, "O") and not is_board_full(board):

            computer_move()

            print("\nComputer placed 'X':")
            print_board(board)

            if is_winner(board, "X"):

                print("Computer wins! Better luck next time.")
                print("Final Path Cost =", path_cost)

                break

        else:
            break

        # Player's turn
        if not is_winner(board, "X"):

            player_move()

            print("\nPlayer placed 'O':")
            print_board(board)

            if is_winner(board, "O"):

                print("Congratulations! You win!")
                print("Final Path Cost =", path_cost)

                break

        if is_board_full(board):

            print("It's a tie!")
            print("Final Path Cost =", path_cost)


play_game()
```

```python
# week1
# Simple Reflex Vacuum Cleaner Agent

rooms = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"

for i in range(4):

    print("\nLocation:", location)
    print("Rooms:", rooms)

    if rooms[location] == "Dirty":
        print("Action: Suck")
        rooms[location] = "Clean"

    elif location == "A":
        print("Action: Move Right")
        location = "B"

    else:
        print("Action: Move Left")
        location = "A"

print("\nFinal State:", rooms)
```

```python
#week1
# Goal Based Vacuum Cleaner Agent
import time

class VacuumCleaner:
    CLEAN = 'Clean'
    DIRTY = 'Dirty'

    def __init__(self):
        # Initialize user input variables
        self.room_a_status = self.DIRTY
        self.room_b_status = self.DIRTY
        self.current_location = 'A'

    def draw_environment(self):
        print(f"Room A: {self.room_a_status} Room B: {self.room_b_status}")

    def scan_and_clean(self, room):
        status = getattr(self, f"room_{room.lower()}_status")
        if status == self.DIRTY:
            setattr(self, f"room_{room.lower()}_status", self.CLEAN)
            return True
        else:
            return False

    def move_left(self):
        self.current_location = 'A'

    def move_right(self):
        self.current_location = 'B'

    def run(self):
        print("Initial environment (Before cleaning):")
        self.draw_environment()
        while any(status == self.DIRTY for status in [self.room_a_status, self.room_b_status]):
            while self.scan_and_clean(self.current_location):
                print(f"\nCleaning in Room {self.current_location}...")
                time.sleep(1)
                self.draw_environment()

            if self.current_location == 'A':
                self.move_right()
            else:
                self.move_left()

            print(f"\nMoving to Room {self.current_location}...")
            time.sleep(1)
            self.draw_environment()

        print("\nFinal environment (After cleaning):")
        self.draw_environment()
        print("\nAll rooms are clean. Task completed.")

if __name__ == "__main__":
    vacuum = VacuumCleaner()
    vacuum.run()



```
